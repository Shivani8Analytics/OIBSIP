# **Step 3 Missing data handling**

# **Missing Data Handling Strategy and Justification**

**1. Categorical Features (No Facility Available)**

**Columns:** Alley, Mas Vnr Type, Bsmt Qual, Bsmt Cond, Bsmt Exposure, BsmtFin Type 1, BsmtFin Type 2, Fireplace Qu, Garage Type, Garage Finish, Garage Qual, Garage Cond, Pool QC, Fence, Misc Feature

**Reason:** In this dataset, missing value means the house does not have that facility. For example, missing value in Pool QC means there is no pool in the house. So we fill these missing values with None.

**2. Numerical Features (No Facility Available)**

**Columns:** Mas Vnr Area, BsmtFin SF 1, BsmtFin SF 2, Bsmt Unf SF, Total Bsmt SF, Bsmt Full Bath, Bsmt Half Bath, Garage Yr Blt, Garage Cars, Garage Area

**Reason:** If a house does not have a basement, garage, or masonry veneer, its area, room count, or year built is logically zero. So we fill these missing values with 0.

**3. Lot Frontage (Continuous Numeric Value)**

**Column:** Lot Frontage

**Reason:** Linear feet of street connected to property varies by location. Using mean can be affected by extreme values. So we fill missing values using the median value of that specific Neighborhood.

**4. Electrical (Single Missing Value)**

**Column:** Electrical

**Reason:** Only one row is missing. So we fill it with the most common electrical system in the dataset, which is mode imputation.

```python
# 1. Fill missing categorical values with 'None' (indicates feature is not present)

none_cols = [
    'Alley', 'Mas Vnr Type', 'Bsmt Qual', 'Bsmt Cond', 'Bsmt Exposure', 
    'BsmtFin Type 1', 'BsmtFin Type 2', 'Fireplace Qu', 'Garage Type', 
    'Garage Finish', 'Garage Qual', 'Garage Cond', 'Pool QC', 'Fence', 'Misc Feature'
]
df[none_cols] = df[none_cols].fillna('None')

# 2. Fill missing numerical values with 0 (indicates zero area, count, or year)

zero_cols = [
    'Mas Vnr Area', 'BsmtFin SF 1', 'BsmtFin SF 2', 'Bsmt Unf SF', 
    'Total Bsmt SF', 'Bsmt Full Bath', 'Bsmt Half Bath', 'Garage Yr Blt', 
    'Garage Cars', 'Garage Area'
]
df[zero_cols] = df[zero_cols].fillna(0)

# 3. Fill Lot Frontage with Neighborhood median, and overall median if neighborhood median is missing

df['Lot Frontage'] = df.groupby('Neighborhood')['Lot Frontage'].transform(lambda x: x.fillna(x.median()))
df['Lot Frontage'] = df['Lot Frontage'].fillna(df['Lot Frontage'].median())

# 4. Fill missing Electrical value with the most frequent value (mode)

df['Electrical'] = df['Electrical'].fillna(df['Electrical'].mode()[0])
```

```python
# Check count of missing values

print("Total missing values left:", df.isnull().sum().sum())
```

**Insights from Missing Data Handling**

**Structural Missing Values (Not Random):** 

1. Most missing values in this dataset (like Pool QC, Garage Finish, Basement Quality) were not missing by error. 

2. They simply represented houses where those features or facilities do not exist.

**Appropriate Imputation:**

1. Replaced missing facility details with None for categories and 0 for numerical measurements (area, counts) to retain feature presence information.

2. Used Neighborhood-based Median for Lot Frontage to handle neighborhood size differences accurately without introducing extreme value bias.

3. Used Mode Imputation for Electrical as it had only a single missing record.

**Result:** All 25 missing columns across 82 total features are now 100% clean, leaving 0 missing values in the dataset without dropping any rows or losing sample size.

# **Step 6 Outlier detection**

```python
# 1. Select numeric columns dynamically (excluding ID and Year columns)
exclude_cols = ['Order', 'PID', 'Year Built', 'Year Remod/Add', 'Garage Yr Blt', 'Yr Sold', 'Mo Sold']
numeric_cols = [col for col in df.select_dtypes(include=['int64', 'float64']).columns if col not in exclude_cols]

# 2. Identify outliers using IQR method and store results
outlier_summary = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Count rows outside lower and upper bounds
    count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
    if count > 0:
        outlier_summary[col] = count

# Display columns and their outlier count
print("Columns with outliers and their counts:")
for col, count in outlier_summary.items():
    print(f"{col}: {count} outliers")
```

# **Outlier Detection Insights and Strategy Decision**

**Detection Insights**

1. We checked for outliers in numerical columns using the IQR (Interquartile Range) method.

2. Large numbers of outliers were found in columns like Enclosed Porch (459), Screen Porch (256), Lot Frontage (206), and SalePrice (137).

3. Some numerical columns represent ratings, codes, or simple room counts (such as MS SubClass, Overall Qual, Overall Cond, Bathrooms, and Fireplaces). Changing extreme values in these columns would change their original meaning.


**Strategy Decision (Selective Capping)**

1. **Decision:** We will cap extreme high and low values using the clip method only on continuous numerical columns (such as SalePrice, Gr Liv Area, Lot Area, Total Bsmt SF, Garage Area, Lot Frontage, Mas Vnr Area, Wood Deck SF, and Open Porch SF).

2. **Exclusion:** We will not cap rating scale columns, room counts, or category code columns.

**Reason:** Capping continuous features keeps extreme house prices and land areas from messing up our model, while keeping all 2,930 rows safe without deleting any data.

```python
# Select only true continuous numerical features for capping
continuous_cols = [
    'SalePrice', 'Gr Liv Area', 'Lot Area', 'Total Bsmt SF', 
    'Garage Area', 'Lot Frontage', 'Mas Vnr Area', 'Wood Deck SF', 
    'Open Porch SF', '1st Flr SF', '2nd Flr SF', 'BsmtFin SF 1'
]

# Apply IQR Capping on continuous features only
for col in continuous_cols:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Cap extreme values
        df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

print("Capping applied successfully on true continuous numerical columns!")
```

# **Final Outlier Handling Summary and Insights**

**Action Taken:** Successfully capped extreme low and high values for all continuous spatial and price columns using the clip method.

**Preservation of Ratings:** Rating columns (like Overall Qual and Overall Cond) and room count columns were left untouched to protect their true category scale.

**Result:** Extreme values in house prices and room areas are controlled without deleting any rows from the dataset. All 2,930 rows remain safe and ready for the next steps.
