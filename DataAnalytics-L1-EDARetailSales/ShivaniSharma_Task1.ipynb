**Step 6 Time series analysis:

Plotting and Analysing  monthly and quarterly sales trends using line charts


**Monthly & Quarterly Sales Aggregation

```python
df_time = df.set_index('Date') # Used for Resampling (Grouping into Monthly/Quarterly)

monthly_sales = df_time['Total Amount'].resample('ME').sum() #grouping data by month based on the Date index and calculates the total sales sum for each month.
quarterly_sales = df_time['Total Amount'].resample('QE').sum() ##grouping data by quarter based on the Date index and calculates the total sales sum for each quarter

print("Monthly Sales Summary\n", monthly_sales)
print("\n Quarterly Sales Summary\n", quarterly_sales)
```

**Plotting monthly and quarterly sales trends using line charts

```python
# Plotting monthly sales trends using line charts

plt.figure(figsize=(10, 4))
plt.plot(monthly_sales.index, monthly_sales.values, marker = 'o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
# Plotting quarterly sales trends using line charts

plt.figure(figsize=(10, 4))
plt.plot(quarterly_sales.index, quarterly_sales.values, marker = 'o')
plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

**Observation for Monthly Sales Trend

1. Monthly Sales mostly stay between 24K and 47K, with no steady long term increase or decrease.
2. There were two main significant strong peaks in the months of May-June(2023) sales bein approx. 53k and in the month of Nov approx. 47K
3. The lowest significant low point was observed in the month of Oct(2023) sales being approx. around 23k-24k.
4. Sales remain healthy till Jan 2024 which is around 42k approx.
5. Feb 2024 shows sudden crash to 1k-2k approx, with no warning signs.
6. This sudden crash/drop breaks the pattern of the previous 13 months, this suggest might be due to incomplete data rather real crash down.

**Observation for Quarterly Sales Trend

1. Mix of growth and decline, obvious from the graph Sales move from 90K to 125K approx across the Quarter.
2. The Q3(2023-07) and Q1(2024-01) has the highest peak, sales being around 124K and 126K approx respectively.

4. Q1 (2024) ends on higher note approx 126k suggesting to be the best quarter so far.
5. Q2 (2024) shows extreme drop to near about 1k-2k.

```python
# Age Group vs. Product Category

# Creating Age groups
bins = [0, 25, 35, 50, 65, 100]
labels = ['18-25', '26-35', '36-50', '51-65', '65+']

# Convert continuous 'Age' numbers into discrete categorical 'Age Groups' (bins)
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

print(df['Age Group'])

# Group data by Age Group and Product Category, count the occurrences, and reshape into a matrix
age_product_pivot = df.groupby(['Age Group', 'Product Category'], observed=False).size().unstack()

print("\n", age_product_pivot)
```

```python
# Heatmap Correlation

sns.heatmap(age_product_pivot.iloc[::-1], annot=True, fmt='d', cmap='coolwarm')
plt.title('Product Category Preference Across Age Groups')
plt.xlabel('Product Category')
plt.ylabel('Age Group')
plt.show()
```

**Observation Summary of Product Category Preference Across Age Groups

1. 36-50 and 51-65 age groups are the biggest buyers overall, with counts significantly higher (88-113) across all three          categories compared to younger groups.

2. Clothing is the most preferred category for both 36-50 (113) and 51-65 (113) age groups — tied as the highest single value      in the entire heatmap.

3. 18-25 age group shows the lowest engagement among active buyers (51-60 range), meaning younger customers purchase noticeably    less than middle-aged and older groups.

4. 65+ age group shows zero purchases across all categories — this is a major red flag, meaning either:
   This age group genuinely doesn't shop in this dataset, or
   Data is missing/not recorded for this segment (similar to the earlier "incomplete data" pattern seen in sales trends)


   Non-obvious insight: Contrary to common assumption that younger customers (18-25) drive the most purchases, it's actually      the 36-65 age range that dominates buying activity — especially in Clothing and Electronics.
